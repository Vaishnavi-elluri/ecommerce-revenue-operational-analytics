### 1. Validate relationships between datasets

```python
# Relationship validation

checks = {
    "Order Items → Orders":
        (~order_items["order_id"].isin(orders["order_id"])).sum(),

    "Order Items → Products":
        (~order_items["product_id"].isin(products["product_id"])).sum(),

    "Order Items → Sellers":
        (~order_items["seller_id"].isin(sellers["seller_id"])).sum(),

    "Orders → Customers":
        (~orders["customer_id"].isin(customers["customer_id"])).sum(),

    "Payments → Orders":
        (~order_payments["order_id"].isin(orders["order_id"])).sum(),

    "Reviews → Orders":
        (~order_reviews["order_id"].isin(orders["order_id"])).sum(),

    "Products → Category Translation":
        (~products["product_category_name"]
         .dropna()
         .isin(category_translation["product_category_name"])).sum()
}

for relationship, missing in checks.items():
    print(f"{relationship}: {missing} missing")
```

```python
order_items["order_value"] = (
    order_items["price"] + order_items["freight_value"]
)

order_items[["price", "freight_value", "order_value"]].head()
```

```python
orders["delivery_days"] = (
    orders["order_delivered_customer_date"]
    - orders["order_purchase_timestamp"]
).dt.days

orders["delivery_delay_days"] = (
    orders["order_delivered_customer_date"]
    - orders["order_estimated_delivery_date"]
).dt.days

orders[
    [
        "order_id",
        "order_status",
        "order_purchase_timestamp",
        "order_delivered_customer_date",
        "order_estimated_delivery_date",
        "delivery_days",
        "delivery_delay_days"
    ]
].head()
```

### Create an On-Time/Late status

```python
orders["delivery_performance"] = orders["delivery_delay_days"].apply(
    lambda x: "Late" if x > 0
    else "On Time" if x == 0
    else "Early" if x < 0
    else "Not Delivered"
)

orders[
    ["order_id", "delivery_delay_days", "delivery_performance"]
].head()
```

```python
order_items["freight_percentage"] = (
    order_items["freight_value"] / order_items["order_value"]
) * 100

order_items[
    ["price", "freight_value", "order_value", "freight_percentage"]
].head()
```

```python
# ============================================================
# 1. ORDER-LEVEL VALUE
# ============================================================

order_value_summary = (
    order_items.groupby("order_id", as_index=False)
    .agg(
        total_order_value=("order_value", "sum"),
        total_freight_value=("freight_value", "sum"),
        total_item_price=("price", "sum"),
        item_count=("order_id", "count")
    )
)

order_value_summary["freight_percentage"] = (
    order_value_summary["total_freight_value"]
    / order_value_summary["total_order_value"]
) * 100


# ============================================================
# 2. ORDER DATE FEATURES
# ============================================================

orders["order_year"] = (
    orders["order_purchase_timestamp"].dt.year
)

orders["order_month"] = (
    orders["order_purchase_timestamp"].dt.month
)

orders["order_month_name"] = (
    orders["order_purchase_timestamp"].dt.strftime("%b")
)

orders["order_year_month"] = (
    orders["order_purchase_timestamp"]
    .dt.to_period("M")
    .astype(str)
)


# ============================================================
# 3. PAYMENT SUMMARY
# ============================================================

payment_summary = (
    order_payments.groupby("order_id", as_index=False)
    .agg(
        total_payment_value=("payment_value", "sum"),
        payment_count=("payment_sequential", "count")
    )
)

payment_type_summary = (
    order_payments.groupby("order_id")["payment_type"]
    .first()
    .reset_index()
    .rename(columns={"payment_type": "primary_payment_type"})
)


# ============================================================
# 4. REVIEW CATEGORY
# ============================================================

def classify_review(score):
    if score >= 4:
        return "Positive"
    elif score == 3:
        return "Neutral"
    else:
        return "Negative"


order_reviews["review_category"] = (
    order_reviews["review_score"]
    .apply(classify_review)
)


# ============================================================
# 5. REVIEW SUMMARY AT ORDER LEVEL
# ============================================================

review_summary = (
    order_reviews.groupby("order_id", as_index=False)
    .agg(
        average_review_score=("review_score", "mean"),
        review_count=("review_id", "count")
    )
)

review_summary["review_category"] = (
    review_summary["average_review_score"]
    .apply(
        lambda x:
        "Positive" if x >= 4
        else "Neutral" if x == 3
        else "Negative"
    )
)


# ============================================================
# 6. DELIVERY INDICATORS
# ============================================================

orders["is_delivered"] = (
    orders["order_status"] == "delivered"
)

orders["is_late"] = (
    orders["delivery_delay_days"] > 0
)

orders["is_early"] = (
    orders["delivery_delay_days"] < 0
)


# ============================================================
# 7. CHECK EVERYTHING
# ============================================================

print("Order Value Summary:")
print(order_value_summary.head())

print("\nPayment Summary:")
print(payment_summary.head())

print("\nPayment Type Summary:")
print(payment_type_summary.head())

print("\nReview Summary:")
print(review_summary.head())

print("\nOrders with derived columns:")
print(
    orders[
        [
            "order_id",
            "order_year",
            "order_month",
            "order_year_month",
            "delivery_days",
            "delivery_delay_days",
            "delivery_performance",
            "is_delivered",
            "is_late",
            "is_early"
        ]
    ].head()
)
```

```python
# ============================================================
# OVERALL E-COMMERCE BUSINESS PERFORMANCE
# ============================================================

total_orders = orders["order_id"].nunique()

delivered_orders = (
    orders["order_status"] == "delivered"
).sum()

total_order_value = (
    order_value_summary["total_order_value"].sum()
)

average_order_value = (
    order_value_summary["total_order_value"].mean()
)

total_freight = (
    order_value_summary["total_freight_value"].sum()
)

average_freight_percentage = (
    order_value_summary["freight_percentage"].mean()
)

late_orders = (
    orders["delivery_performance"] == "Late"
).sum()

early_orders = (
    orders["delivery_performance"] == "Early"
).sum()

on_time_orders = (
    orders["delivery_performance"] == "On Time"
).sum()

average_delivery_days = (
    orders.loc[
        orders["delivery_days"].notna(),
        "delivery_days"
    ].mean()
)

average_review_score = (
    order_reviews["review_score"].mean()
)


# ============================================================
# DISPLAY
# ============================================================

print("Total Orders:", total_orders)
print("Delivered Orders:", delivered_orders)
print("Total Order Value:", round(total_order_value, 2))
print("Average Order Value:", round(average_order_value, 2))
print("Total Freight:", round(total_freight, 2))
print("Average Freight %:", round(average_freight_percentage, 2))
print("Late Orders:", late_orders)
print("Early Orders:", early_orders)
print("On-Time Orders:", on_time_orders)
print("Average Delivery Days:", round(average_delivery_days, 2))
print("Average Review Score:", round(average_review_score, 2))
```

```python
# ============================================================
# MONTHLY ORDER VALUE & ORDER VOLUME
# ============================================================

monthly_performance = (
    orders[["order_id", "order_year_month"]]
    .merge(
        order_value_summary[
            ["order_id", "total_order_value"]
        ],
        on="order_id",
        how="left"
    )
    .groupby("order_year_month", as_index=False)
    .agg(
        total_orders=("order_id", "nunique"),
        total_order_value=("total_order_value", "sum"),
        average_order_value=("total_order_value", "mean")
    )
    .sort_values("order_year_month")
)


# ============================================================
# HIGHEST / LOWEST MONTHS
# ============================================================

highest_value_month = monthly_performance.loc[
    monthly_performance["total_order_value"].idxmax()
]

lowest_value_month = monthly_performance.loc[
    monthly_performance["total_order_value"].idxmin()
]

highest_order_volume_month = monthly_performance.loc[
    monthly_performance["total_orders"].idxmax()
]

lowest_order_volume_month = monthly_performance.loc[
    monthly_performance["total_orders"].idxmin()
]


# ============================================================
# DISPLAY
# ============================================================

print("MONTHLY PERFORMANCE")
print(monthly_performance)

print("\nHighest Order Value Month:")
print(highest_value_month)

print("\nLowest Order Value Month:")
print(lowest_value_month)

print("\nHighest Order Volume Month:")
print(highest_order_volume_month)

print("\nLowest Order Volume Month:")
print(lowest_order_volume_month)
```

```python
# ============================================================
# FILTER TO MEANINGFUL / COMPLETE ANALYSIS PERIOD
# ============================================================

monthly_analysis = monthly_performance[
    monthly_performance["order_year_month"].between(
        "2017-01", "2018-08"
    )
].copy()


# ============================================================
# YEARLY PERFORMANCE
# ============================================================

orders["order_year"] = (
    orders["order_purchase_timestamp"].dt.year
)

yearly_performance = (
    orders[orders["order_year"].isin([2017, 2018])]
    [["order_id", "order_year"]]
    .merge(
        order_value_summary[
            ["order_id", "total_order_value"]
        ],
        on="order_id",
        how="left"
    )
    .groupby("order_year", as_index=False)
    .agg(
        total_orders=("order_id", "nunique"),
        total_order_value=("total_order_value", "sum"),
        average_order_value=("total_order_value", "mean")
    )
)


# ============================================================
# 2017 vs 2018 CHANGE
# ============================================================

value_2017 = yearly_performance.loc[
    yearly_performance["order_year"] == 2017,
    "total_order_value"
].iloc[0]

value_2018 = yearly_performance.loc[
    yearly_performance["order_year"] == 2018,
    "total_order_value"
].iloc[0]

orders_2017 = yearly_performance.loc[
    yearly_performance["order_year"] == 2017,
    "total_orders"
].iloc[0]

orders_2018 = yearly_performance.loc[
    yearly_performance["order_year"] == 2018,
    "total_orders"
].iloc[0]

value_growth = (
    (value_2018 - value_2017) / value_2017
) * 100

order_growth = (
    (orders_2018 - orders_2017) / orders_2017
) * 100


# ============================================================
# DISPLAY
# ============================================================

print("MONTHLY ANALYSIS PERIOD")
print(monthly_analysis)

print("\nYEARLY PERFORMANCE")
print(yearly_performance)

print("\n2017 → 2018 Order Value Change:",
      round(value_growth, 2), "%")

print("2017 → 2018 Order Volume Change:",
      round(order_growth, 2), "%")
```

```python
# ============================================================
# PRODUCT CATEGORY PERFORMANCE
# ============================================================

category_performance = (
    order_items
    .merge(
        products[
            ["product_id", "product_category_name"]
        ],
        on="product_id",
        how="left"
    )
    .merge(
        category_translation,
        on="product_category_name",
        how="left"
    )
)


# ============================================================
# USE ENGLISH CATEGORY NAME WHERE AVAILABLE
# OTHERWISE KEEP ORIGINAL CATEGORY NAME
# ============================================================

category_performance["category"] = (
    category_performance["product_category_name_english"]
    .fillna(category_performance["product_category_name"])
)


# ============================================================
# AGGREGATE CATEGORY METRICS
# ============================================================

category_summary = (
    category_performance
    .groupby("category", as_index=False)
    .agg(
        total_items=("order_item_id", "count"),
        total_order_value=("order_value", "sum"),
        total_freight=("freight_value", "sum"),
        average_item_price=("price", "mean"),
        unique_orders=("order_id", "nunique")
    )
)


# ============================================================
# ADD CATEGORY SHARE
# ============================================================

category_summary["order_value_share"] = (
    category_summary["total_order_value"]
    / category_summary["total_order_value"].sum()
) * 100


# ============================================================
# TOP 10 CATEGORIES BY ORDER VALUE
# ============================================================

top_categories = (
    category_summary
    .sort_values("total_order_value", ascending=False)
    .head(10)
)


print("TOP 10 CATEGORIES BY ORDER VALUE")
print(top_categories)


# ============================================================
# TOP 10 CATEGORIES BY NUMBER OF ITEMS
# ============================================================

top_volume_categories = (
    category_summary
    .sort_values("total_items", ascending=False)
    .head(10)
)

print("\nTOP 10 CATEGORIES BY ITEM VOLUME")
print(top_volume_categories)
```

```python
# ============================================================
# SELLER PERFORMANCE
# ============================================================

seller_performance = (
    order_items
    .merge(
        sellers[
            ["seller_id", "seller_zip_code_prefix",
             "seller_city", "seller_state"]
        ],
        on="seller_id",
        how="left"
    )
)


# ============================================================
# SELLER SUMMARY
# ============================================================

seller_summary = (
    seller_performance
    .groupby(
        [
            "seller_id",
            "seller_city",
            "seller_state"
        ],
        as_index=False
    )
    .agg(
        total_items=("order_item_id", "count"),
        unique_orders=("order_id", "nunique"),
        total_order_value=("order_value", "sum"),
        total_freight=("freight_value", "sum"),
        average_item_price=("price", "mean")
    )
)


# ============================================================
# ORDER VALUE SHARE
# ============================================================

seller_summary["order_value_share"] = (
    seller_summary["total_order_value"]
    / seller_summary["total_order_value"].sum()
) * 100


# ============================================================
# TOP 10 SELLERS BY ORDER VALUE
# ============================================================

top_sellers_value = (
    seller_summary
    .sort_values(
        "total_order_value",
        ascending=False
    )
    .head(10)
)

print("TOP 10 SELLERS BY ORDER VALUE")
print(top_sellers_value)


# ============================================================
# TOP 10 SELLERS BY ITEM VOLUME
# ============================================================

top_sellers_volume = (
    seller_summary
    .sort_values(
        "total_items",
        ascending=False
    )
    .head(10)
)

print("\nTOP 10 SELLERS BY ITEM VOLUME")
print(top_sellers_volume)
```

```python
# ============================================================
# DELIVERY PERFORMANCE ANALYSIS
# ============================================================

delivery_summary = (
    orders["delivery_performance"]
    .value_counts()
    .reset_index()
)

delivery_summary.columns = [
    "delivery_performance",
    "order_count"
]

delivery_summary["percentage"] = (
    delivery_summary["order_count"]
    / delivery_summary["order_count"].sum()
) * 100


# ============================================================
# AVERAGE DELIVERY METRICS
# ============================================================

avg_delivery_days = orders["delivery_days"].mean()

avg_delay_days = orders["delivery_delay_days"].mean()

late_orders = orders[
    orders["delivery_performance"] == "Late"
]

average_late_delay = late_orders[
    "delivery_delay_days"
].mean()


# ============================================================
# LATE ORDERS BY YEAR
# ============================================================

late_by_year = (
    orders[orders["delivery_performance"] == "Late"]
    .groupby("order_year")
    .size()
    .reset_index(name="late_orders")
)


# ============================================================
# DISPLAY
# ============================================================

print("DELIVERY PERFORMANCE")
print(delivery_summary)

print("\nAverage Delivery Days:",
      round(avg_delivery_days, 2))

print("Average Delivery Delay:",
      round(avg_delay_days, 2))

print("Average Late Delivery Delay:",
      round(average_late_delay, 2))

print("\nLATE ORDERS BY YEAR")
print(late_by_year)
```

```python
# ============================================================
# DELIVERY PERFORMANCE vs CUSTOMER REVIEWS
# ============================================================

delivery_reviews = (
    orders[
        [
            "order_id",
            "delivery_performance",
            "delivery_days",
            "delivery_delay_days"
        ]
    ]
    .merge(
        review_summary[
            [
                "order_id",
                "average_review_score"
            ]
        ],
        on="order_id",
        how="inner"
    )
)


# ============================================================
# AVERAGE REVIEW SCORE BY DELIVERY PERFORMANCE
# ============================================================

review_by_delivery = (
    delivery_reviews
    .groupby("delivery_performance", as_index=False)
    .agg(
        orders=("order_id", "nunique"),
        average_review_score=("average_review_score", "mean")
    )
    .sort_values(
        "average_review_score",
        ascending=False
    )
)


# ============================================================
# REVIEW SCORE BY LATE / NOT LATE
# ============================================================

delivery_reviews["late_flag"] = (
    delivery_reviews["delivery_performance"] == "Late"
)

review_by_late_flag = (
    delivery_reviews
    .groupby("late_flag", as_index=False)
    .agg(
        orders=("order_id", "nunique"),
        average_review_score=("average_review_score", "mean")
    )
)


# ============================================================
# DISPLAY
# ============================================================

print("REVIEW SCORE BY DELIVERY PERFORMANCE")
print(review_by_delivery)

print("\nREVIEW SCORE: LATE vs NOT LATE")
print(review_by_late_flag)
```

```python
# ============================================================
# DELIVERY PERFORMANCE BY PRODUCT CATEGORY
# ============================================================

category_delivery = (
    order_items[
        [
            "order_id",
            "product_id"
        ]
    ]
    .merge(
        category_performance[
            [
                "product_id",
                "category"
            ]
        ].drop_duplicates("product_id"),
        on="product_id",
        how="left"
    )
    .merge(
        orders[
            [
                "order_id",
                "delivery_performance",
                "delivery_delay_days"
            ]
        ],
        on="order_id",
        how="left"
    )
)


# ============================================================
# CATEGORY DELIVERY SUMMARY
# ============================================================

category_delivery_summary = (
    category_delivery
    .groupby("category", as_index=False)
    .agg(
        total_items=("product_id", "count"),
        late_items=(
            "delivery_performance",
            lambda x: (x == "Late").sum()
        ),
        average_delay_days=(
            "delivery_delay_days",
            "mean"
        )
    )
)


# ============================================================
# LATE DELIVERY RATE
# ============================================================

category_delivery_summary["late_rate"] = (
    category_delivery_summary["late_items"]
    / category_delivery_summary["total_items"]
) * 100


# ============================================================
# FILTER OUT VERY SMALL CATEGORIES
# ============================================================

category_delivery_summary_filtered = (
    category_delivery_summary[
        category_delivery_summary["total_items"] >= 100
    ]
    .sort_values(
        "late_rate",
        ascending=False
    )
)


# ============================================================
# TOP CATEGORIES WITH HIGHEST LATE RATE
# ============================================================

print("CATEGORIES WITH HIGHEST LATE DELIVERY RATE")
print(
    category_delivery_summary_filtered.head(15)
)
```

```python
# ============================================================
# CATEGORY-LEVEL DELIVERY PERFORMANCE
# ORDER-LEVEL METRICS
# ============================================================

category_order_delivery = (
    order_items[
        ["order_id", "product_id"]
    ]
    .merge(
        products[
            ["product_id", "product_category_name"]
        ],
        on="product_id",
        how="left"
    )
    .merge(
        category_translation,
        on="product_category_name",
        how="left"
    )
)

# English category where available
category_order_delivery["category"] = (
    category_order_delivery["product_category_name_english"]
    .fillna(
        category_order_delivery["product_category_name"]
    )
)

# One category per order-item
category_order_delivery = (
    category_order_delivery[
        ["order_id", "category"]
    ]
    .drop_duplicates()
)

# Add delivery information
category_order_delivery = (
    category_order_delivery
    .merge(
        orders[
            [
                "order_id",
                "delivery_performance",
                "delivery_delay_days"
            ]
        ],
        on="order_id",
        how="left"
    )
)


# ============================================================
# CATEGORY DELIVERY SUMMARY
# ============================================================

category_delivery_final = (
    category_order_delivery
    .groupby("category", as_index=False)
    .agg(
        total_orders=("order_id", "nunique"),
        late_orders=(
            "delivery_performance",
            lambda x: (x == "Late").sum()
        ),
        average_delivery_delay=(
            "delivery_delay_days",
            "mean"
        )
    )
)

# Late delivery rate
category_delivery_final["late_rate"] = (
    category_delivery_final["late_orders"]
    / category_delivery_final["total_orders"]
) * 100


# ============================================================
# REMOVE VERY SMALL CATEGORIES
# ============================================================

category_delivery_final = (
    category_delivery_final[
        category_delivery_final["total_orders"] >= 100
    ]
    .sort_values(
        "late_rate",
        ascending=False
    )
)


# ============================================================
# TOP CATEGORIES BY LATE RATE
# ============================================================

print("TOP CATEGORIES BY ORDER-LEVEL LATE RATE")
print(
    category_delivery_final.head(15)
)
```

```python
# ============================================================
# PAYMENT METHOD ANALYSIS
# ============================================================

payment_analysis = (
    order_payments
    .groupby("payment_type", as_index=False)
    .agg(
        total_transactions=("order_id", "count"),
        unique_orders=("order_id", "nunique"),
        total_payment_value=("payment_value", "sum"),
        average_payment_value=("payment_value", "mean")
    )
)

# Payment share by transaction count
payment_analysis["transaction_share"] = (
    payment_analysis["total_transactions"]
    / payment_analysis["total_transactions"].sum()
) * 100

# Payment share by value
payment_analysis["value_share"] = (
    payment_analysis["total_payment_value"]
    / payment_analysis["total_payment_value"].sum()
) * 100

payment_analysis = payment_analysis.sort_values(
    "total_payment_value",
    ascending=False
)

print("PAYMENT METHOD ANALYSIS")
print(payment_analysis)
```
